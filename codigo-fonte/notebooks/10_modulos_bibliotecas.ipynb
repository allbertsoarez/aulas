# 📦 Módulos e Bibliotecas

O Python tem uma filosofia: **"Batteries included"** (baterias inclusas). Isso significa que ele já vem com centenas de módulos prontos para usar.

Além disso, existe o **PyPI** (Python Package Index), com mais de 500.000 pacotes criados pela comunidade.

### 🏛️ Módulos da Biblioteca Padrão (já vêm com o Python)

| Módulo | Função |
| :--- | :--- |
| `math` | Funções matemáticas (raiz, trigonometria, etc.) |
| `random` | Geração de números aleatórios |
| `datetime` | Manipulação de datas e horas |
| `os` | Interação com o sistema operacional |
| `json` | Leitura e escrita de arquivos JSON |
| `csv` | Manipulação de arquivos CSV |

````python
# 📐 Módulo math
import math

print(f"Valor de PI: {math.pi}")
print(f"Raiz quadrada de 16: {math.sqrt(16)}")
print(f"2 elevado a 10: {math.pow(2, 10)}")
print(f"Seno de 90°: {math.sin(math.radians(90))}")
````

````python
# 🎲 Módulo random
import random

# Número aleatório entre 1 e 10
print(f"Número aleatório: {random.randint(1, 10)}")

# Escolher um item aleatório de uma lista
frutas = ["Maçã", "Banana", "Laranja", "Uva"]
print(f"Fruta sorteada: {random.choice(frutas)}")

# Embaralhar uma lista
random.shuffle(frutas)
print(f"Lista embaralhada: {frutas}")
````

````python
# 📅 Módulo datetime
from datetime import datetime, timedelta

agora = datetime.now()
print(f"Agora: {agora}")
print(f"Ano: {agora.year}, Mês: {agora.month}, Dia: {agora.day}")

# Somar 7 dias à data atual
semana_que_vem = agora + timedelta(days=7)
print(f"Daqui a 7 dias: {semana_que_vem.strftime('%d/%m/%Y')}")
````

````python
# 📋 Módulo json (muito usado em APIs)
import json

# Converter dicionário Python para JSON (serialização)
aluno = {
    "nome": "Ana Silva",
    "idade": 20,
    "cursos": ["Python", "Banco de Dados"]
}

json_string = json.dumps(aluno, indent=4, ensure_ascii=False)
print("=== JSON ===")
print(json_string)

# Converter JSON de volta para dicionário Python (desserialização)
aluno_de_volta = json.loads(json_string)
print(f"\n=== De volta ao Python ===")
print(f"Nome: {aluno_de_volta['nome']}")
````

### 🌐 Bibliotecas Externas (PyPI)

Para instalar bibliotecas externas, usamos o **pip** (Python Package Installer).

No terminal:
```bash
pip install nome_da_biblioteca

````python
# 🌐 Biblioteca requests (precisa instalar primeiro)
# No Colab, descomente a linha abaixo:
# !pip install requests

import requests

# Consumindo uma API pública (via CEP)
cep = "01001000"  # Praça da Sé, São Paulo
url = f"https://viacep.com.br/ws/{cep}/json/"

resposta = requests.get(url)

if resposta.status_code == 200:
    dados = resposta.json()
    print(f"📍 CEP: {dados['cep']}")
    print(f"🏙️ Logradouro: {dados.get('logradouro', 'N/A')}")
    print(f"🏘️ Bairro: {dados['bairro']}")
    print(f"🌆 Cidade: {dados['localidade']}/{dados['uf']}")
else:
    print(f"❌ Erro: {resposta.status_code}")
````

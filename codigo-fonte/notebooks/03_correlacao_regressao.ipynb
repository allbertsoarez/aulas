  # 🔗 Correlação e Regressão Linear
  
  Vamos investigar se existe alguma relação linear entre o **Ano de Publicação** e o **Preço** dos livros.
  *Lembre-se: Correlação não implica causalidade!*

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# Dados
anos = [1899, 1949, 1977, 1986, 1997, 2020]
precos = [29.90, 39.90, 24.90, 69.90, 49.90, 89.90]

# Calculando o Coeficiente de Correlação de Pearson (r)
correlacao, p_valor = stats.pearsonr(anos, precos)
print(f"Coeficiente de Correlação (r): {correlacao:.3f}")
print(f"P-valor: {p_valor:.4f}")

if p_valor < 0.05:
    print("✅ Há evidência estatística de correlação linear (p < 0.05).")
else:
    print("❌ Não há evidência estatística suficiente de correlação linear.")
```

```python
# Gráfico de Dispersão com Linha de Regressão (Tendência)
plt.figure(figsize=(8, 5))
sns.regplot(x=anos, y=precos, scatter_kws={'s':100}, line_kws={'color':'red'})

plt.title('Relação entre Ano de Publicação e Preço')
plt.xlabel('Ano de Publicação')
plt.ylabel('Preço (R$)')
plt.show()
```

  # 🗃️ Listas e Dicionários
  - **Listas:** Coleção ordenada de itens (usam índice numérico).
  - **Dicionários:** Coleção de pares Chave-Valor (usam uma "chave" para acessar o dado).

```python
# --- LISTAS ---
frutas = ["Maçã", "Banana", "Laranja"]
frutas.append("Uva")  # Adiciona ao final da lista

print(f"Lista de frutas: {frutas}")
print(f"A primeira fruta é: {frutas[0]}")  # Índices começam em 0!

# Iterando sobre uma lista
print("\nFrutas disponíveis:")
for fruta in frutas:
  print(f"- {fruta}")
```

```python
# --- DICIONÁRIOS ---
aluno = {
    "nome": "Ana Silva",
    "idade": 20,
    "curso": "Ciência da Computação",
    "notas": [8.5, 9.0, 7.5]  # Podemos ter listas dentro de dicionários!
    }

print(f"Dados do aluno: {aluno}")
print(f"Nome do aluno: {aluno['nome']}")
print(f"Média das notas: {sum(aluno['notas']) / len(aluno['notas'])}")

# Adicionando uma nova chave
aluno["status"] = "Ativo"
```

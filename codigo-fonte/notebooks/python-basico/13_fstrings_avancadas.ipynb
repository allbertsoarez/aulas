  # 🎨 f-strings Avançadas: Formatação Profissional
  
  Até agora vimos f-strings básicas: `f"Olá, {nome}"`. Mas as f-strings são **muito mais poderosas**!
  
  Neste notebook vamos aprender a:
  - Formatar números com casas decimais específicas
  - Criar tabelas alinhadas profissionalmente
  - Formatar moedas, porcentagens e datas
  - Fazer debug rápido com `=`
  
  💡 **Por que isso importa?** Em Ciência de Dados e relatórios, a apresentação dos números é tão importante quanto os números em si!

````python
# 📐 Formatação de Números com Casas Decimais

pi = 3.141592653589793
resultado = 1234.56789

print("=== Controle de Casas Decimais ===")
print(f"Pi com 2 casas: {pi:.2f}")      # 3.14
print(f"Pi com 4 casas: {pi:.4f}")      # 3.1416
print(f"Resultado: {resultado:.3f}")    # 1234.568 (arredonda!)

print("\n=== Notação Científica ===")
numero_grande = 1234567890
print(f"Notação científica: {numero_grande:.2e}")  # 1.23e+09
````

````python
# 💰 Separadores de Milhar e Moeda

salario = 1234567.89
preco = 99.9

print("=== Separador de Milhar ===")
print(f"Sem separador: {salario}")           # 1234567.89
print(f"Com vírgula: {salario:,.2f}")        # 1,234,567.89 (padrão americano)

# Para padrão brasileiro (ponto como separador de milhar)
print(f"Padrão BR: {salario:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))

print("\n=== Moeda ===")
print(f"R$ {preco:.2f}")                     # R$ 99.90
print(f"Total: R$ {salario:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))

# Dica: para formatação monetária profissional, use a biblioteca 'locale'
````

````python
# 📊 Porcentagens

taxa_sucesso = 0.8567
crescimento = 0.1234
desconto = 0.05

print("=== Formatação de Porcentagens ===")
print(f"Taxa de sucesso: {taxa_sucesso:.1%}")    # 85.7%
print(f"Crescimento: {taxa_sucesso:.2%}")        # 85.67%
print(f"Desconto: {desconto:.0%}")               # 5%

print("\n=== Caso Real: Relatório de Vendas ===")
vendas_total = 150000
vendas_atingidas = 127500
percentual = vendas_atingidas / vendas_total
print(f"Meta: R$ {vendas_total:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
print(f"Atingido: R$ {vendas_atingidas:,.2f}".replace(",", "X").replace(".", ",").replace("X", "."))
print(f"Progresso: {percentual:.1%}")
````

````python
# 📏 Alinhamento e Preenchimento (Criar Tabelas!)

produtos = [
    ("Notebook Dell", 4500.00),
    ("Mouse Logitech", 150.90),
    ("Teclado Mecânico", 350.00),
    ("Monitor 27'", 1899.99)
    ]

print("=== Tabela Simples ===")
print(f"{'Produto':<20} | {'Preço':>10}")
print("-" * 33)
for nome, preco in produtos:
  print(f"{nome:<20} | R$ {preco:>8.2f}")

print("\n=== Tabela com Preenchimento ===")
print(f"{'Produto':*<20} | {'Preço':*>10}")
print("-" * 33)
for nome, preco in produtos:
  print(f"{nome:*<20} | R$ {preco:*>8.2f}")

print("\n=== Centralizado ===")
print(f"{'RELATÓRIO DE VENDAS':^40}")
print("=" * 40)
````

````python
# 📅 Formatação de Datas

from datetime import datetime

agora = datetime.now()

print("=== Formatos de Data ===")
print(f"Data completa: {agora:%d/%m/%Y %H:%M:%S}")
print(f"Só data: {agora:%d/%m/%Y}")
print(f"Só hora: {agora:%H:%M}")
print(f"Dia da semana: {agora:%A}")
print(f"Mês: {agora:%B}")

print("\n=== Formato ISO ===")
print(f"ISO: {agora:%Y-%m-%d}")

print("\n=== Caso Real: Log de Sistema ===")
print(f"[{agora:%Y-%m-%d %H:%M:%S}] INFO: Sistema iniciado com sucesso")
print(f"[{agora:%Y-%m-%d %H:%M:%S}] ERROR: Falha na conexão com banco de dados")
````

````python
# 🧮 Expressões Dentro das Chaves

a = 10
b = 3

print("=== Expressões Diretas ===")
print(f"Soma: {a + b}")              # 13
print(f"Multiplicação: {a * b}")     # 30
print(f"Divisão: {a / b:.2f}")       # 3.33

# Chamadas de métodos
nome = "ana silva"
print(f"Nome formatado: {nome.title()}")  # Ana Silva

# Condicionais (operador ternário)
idade = 20
print(f"Status: {'Maior de idade' if idade >= 18 else 'Menor de idade'}")

# Acessando índices e chaves
lista = [10, 20, 30, 40, 50]
dicionario = {"nome": "Ana", "idade": 20}
print(f"Primeiro elemento: {lista[0]}")
print(f"Nome do dict: {dicionario['nome']}")
````

````python
# 🐛 Debug Rápido com = (Python 3.8+)

# Em vez de:
x = 10
y = 20
print(f"x = {x}, y = {y}")

# Use (mais rápido e menos propenso a erros):
print(f"{x = }")
print(f"{y = }")
print(f"{x + y = }")

# Útil para debugar expressões complexas
temperatura = 25.5
umidade = 60
print(f"\n{temperatura = :.1f}°C")
print(f"{umidade = }%")

# Debugar listas e dicionários
numeros = [1, 2, 3, 4, 5]
print(f"\n{numeros = }")
print(f"{len(numeros) = }")
print(f"{sum(numeros) = }")
````

````python
# 🎯 CASO DE USO REAL: Relatório Estatístico Profissional

import statistics

# Dados de exemplo (preços de livros da Editora)
precos = [29.90, 39.90, 49.90, 69.90, 24.90, 89.90, 35.00, 42.50, 55.00, 75.00]

# Calculando estatísticas
media = statistics.mean(precos)
mediana = statistics.median(precos)
desvio = statistics.stdev(precos)
minimo = min(precos)
maximo = max(precos)

# Gerando relatório formatado
print("=" * 50)
print(f"{'RELATÓRIO ESTATÍSTICO - PREÇOS DE LIVROS':^50}")
print("=" * 50)
print()
print(f"{'Métrica':<25} | {'Valor':>15}")
print("-" * 43)
print(f"{'Número de observações':<25} | {len(precos):>15}")
print(f"{'Média':<25} | R$ {media:>12.2f}")
print(f"{'Mediana':<25} | R$ {mediana:>12.2f}")
print(f"{'Desvio Padrão':<25} | R$ {desvio:>12.2f}")
print(f"{'Mínimo':<25} | R$ {minimo:>12.2f}")
print(f"{'Máximo':<25} | R$ {maximo:>12.2f}")
print(f"{'Amplitude':<25} | R$ {maximo - minimo:>12.2f}")
print()
print("=" * 50)
print(f"{'Coeficiente de Variação':<25} | {(desvio/media)*100:>14.1f}%")
print("=" * 50)
````

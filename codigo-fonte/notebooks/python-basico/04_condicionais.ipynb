  # 🔀 Estruturas Condicionais
  Permitem que o programa tome decisões.
  **Atenção à indentação (espaços no início da linha):** Em Python, ela é obrigatória e define o que está dentro do `if`.

```python
nota = float(input("Digite a sua nota (0 a 100): "))

if nota >= 70:
  print("✅ Você foi APROVADO!")
elif nota >= 50:
  print("⚠️ Você está de RECUPERAÇÃO.")
else:
  print("❌ Você foi REPROVADO.")

print("Fim do programa.")
```

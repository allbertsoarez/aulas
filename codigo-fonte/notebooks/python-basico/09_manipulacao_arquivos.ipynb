  # 📁 Manipulação de Arquivos
  
  Até agora, nossos dados "morriam" quando o notebook era fechado. Para **persistir dados**, precisamos salvá-los em arquivos.
  
  Em Python, usamos a função `open()` com diferentes modos:
  
  | Modo | Descrição |
  | :---: | :--- |
  | `'r'` | Leitura (padrão). Erro se o arquivo não existir. |
  | `'w'` | Escrita. **Sobrescreve** o arquivo se já existir. |
  | `'a'` | Append (adicionar ao final). |
  | `'r+'` | Leitura e escrita. |

```python
# 📝 ESCREVENDO em um arquivo
with open("meu_arquivo.txt", "w", encoding="utf-8") as arquivo:
  arquivo.write("Olá, Mundo!\n")
  arquivo.write("Esta é a segunda linha.\n")
  arquivo.write("Python é incrível!\n")

  print("✅ Arquivo criado com sucesso!")

# O 'with' garante que o arquivo é fechado automaticamente,
# mesmo se ocorrer um erro.
```

```python
# 📖 LENDO um arquivo (linha por linha)
with open("meu_arquivo.txt", "r", encoding="utf-8") as arquivo:
  for linha in arquivo:
    print(linha.strip())  # strip() remove a quebra de linha extra

    print("\n✅ Leitura concluída!")
```

```python
# 📖 LENDO tudo de uma vez
with open("meu_arquivo.txt", "r", encoding="utf-8") as arquivo:
  conteudo = arquivo.read()
  print("=== Conteúdo completo ===")
  print(conteudo)

# Ou lendo como lista de linhas
with open("meu_arquivo.txt", "r", encoding="utf-8") as arquivo:
  linhas = arquivo.readlines()
  print(f"Total de linhas: {len(linhas)}")
  print(f"Primeira linha: {linhas[0].strip()}")
```

```python
# ➕ ADICIONANDO conteúdo ao final (append)
with open("meu_arquivo.txt", "a", encoding="utf-8") as arquivo:
  arquivo.write("Esta linha foi adicionada depois!\n")

# Verificando o resultado
with open("meu_arquivo.txt", "r", encoding="utf-8") as arquivo:
  print(arquivo.read())
```

  ### 📊 Trabalhando com CSV (Comma-Separated Values)
  
  CSV é o formato mais comum para troca de dados tabulares. Vamos usar o módulo `csv` do Python.

```python
import csv

# 📝 Criando um arquivo CSV
dados = [
    ["Nome", "Idade", "Curso"],
    ["Ana", 20, "Ciência da Computação"],
    ["Bruno", 22, "Engenharia"],
    ["Carla", 19, "Matemática"]
    ]

with open("alunos.csv", "w", encoding="utf-8", newline="") as arquivo:
  escritor = csv.writer(arquivo)
  escritor.writerows(dados)

print("✅ CSV criado!")

# 📖 Lendo o CSV
with open("alunos.csv", "r", encoding="utf-8") as arquivo:
  leitor = csv.reader(arquivo)
  for linha in leitor:
    print(linha)
```

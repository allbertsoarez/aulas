  # 🛡️ Tratamento de Erros (try/except)
  
  Programas reais **quebram**. O usuário digita texto quando você esperava número, um arquivo não existe, a internet cai...
  
  Em vez de deixar o programa travar com uma mensagem feia de erro, podemos **capturar** o erro e tratá-lo de forma elegante.

```python
# ❌ SEM tratamento de erros: o programa queia feio
# numero = int(input("Digite um número: "))  # Se digitar "abc", dá ValueError!
# resultado = 10 / numero                      # Se digitar 0, dá ZeroDivisionError!

# ✅ COM tratamento de erros: o programa continua funcionando
try:
  numero = int(input("Digite um número: "))
  resultado = 10 / numero
  print(f"10 dividido por {numero} é {resultado}")
except ValueError:
  print("❌ Erro: Você não digitou um número inteiro válido!")
except ZeroDivisionError:
  print("❌ Erro: Não é possível dividir por zero!")
except Exception as e:
  print(f"❌ Erro inesperado: {e}")
finally:
  print("✅ Este bloco SEMPRE executa, independente de erro ou não.")
```

  ### 📚 Principais tipos de exceções
  
  | Exceção | Quando ocorre |
  | :--- | :--- |
  | `ValueError` | Valor inválido (ex: converter "abc" para int) |
  | `TypeError` | Operação entre tipos incompatíveis |
  | `ZeroDivisionError` | Divisão por zero |
  | `IndexError` | Índice fora do range de uma lista |
  | `KeyError` | Chave inexistente em um dicionário |
  | `FileNotFoundError` | Arquivo não encontrado |
  | `KeyboardInterrupt` | Usuário pressionou Ctrl+C |

```python
# Exemplo prático: calculadora segura
def dividir():
  try:
    a = float(input("Digite o numerador: "))
    b = float(input("Digite o denominador: "))
    resultado = a / b
  except ValueError:
    print("❌ Você precisa digitar números!")
    return
  except ZeroDivisionError:
    print("❌ Não pode dividir por zero!")
    return
  except KeyboardInterrupt:
    print("\n⚠️ Operação cancelada pelo usuário.")
    return

    print(f"✅ Resultado: {a} / {b} = {resultado}")
    dividir()
```

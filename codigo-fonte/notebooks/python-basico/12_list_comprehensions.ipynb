# ✨ List Comprehensions

**List Comprehensions** são uma forma **concisa e elegante** de criar listas em Python. É uma das características mais "pythonicas" da linguagem.

Em vez de:
```python
lista = []
for i in range(10):
    lista.append(i * 2)

lista = [i * 2 for i in range(10)]

````python
# Exemplos básicos

# Quadrados de 0 a 9
quadrados = [x**2 for x in range(10)]
print(f"Quadrados: {quadrados}")

# Números pares de 0 a 20
pares = [x for x in range(21) if x % 2 == 0]
print(f"Pares: {pares}")

# Converter nomes para maiúsculas
nomes = ["ana", "bruno", "carla"]
nomes_maiusculos = [nome.upper() for nome in nomes]
print(f"Nomes: {nomes_maiusculos}")
````

````python
# Comprehensions com condicional (if/else)

# Classificar números como "par" ou "ímpar"
numeros = range(1, 11)
classificacao = ["par" if x % 2 == 0 else "ímpar" for x in numeros]
print(f"Classificação: {classificacao}")

# Aplicar desconto apenas em produtos acima de R$ 100
precos = [50, 150, 80, 200, 30]
precos_com_desconto = [p * 0.9 if p > 100 else p for p in precos]
print(f"Preços com desconto: {precos_com_desconto}")
````

````python
# Comprehensions aninhadas (matrizes)

# Criar uma matriz 3x3
matriz = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print("Matriz 3x3:")
for linha in matriz:
    print(linha)

# "Achatando" uma matriz (transformar em lista plana)
matriz_original = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
plana = [num for linha in matriz_original for num in linha]
print(f"\nMatriz plana: {plana}")
````

````python
# Dict Comprehensions (para dicionários!)

# Criar dicionário de quadrados
quadrados_dict = {x: x**2 for x in range(1, 6)}
print(f"Dicionário de quadrados: {quadrados_dict}")

# Inverter chaves e valores
original = {"a": 1, "b": 2, "c": 3}
invertido = {v: k for k, v in original.items()}
print(f"Invertido: {invertido}")

# Filtrar dicionário
precos = {"maçã": 5, "banana": 3, "manga": 8, "uva": 10}
caros = {fruta: preco for fruta, preco in precos.items() if preco > 5}
print(f"Frutas caras: {caros}")
````

````python
# Set Comprehensions (para conjuntos - sem duplicatas!)

# Extrair vogais únicas de uma frase
frase = "programação é incrível"
vogais = {letra for letra in frase if letra in "aeiou"}
print(f"Vogais únicas: {vogais}")

# Encontrar números comuns entre duas listas
lista1 = [1, 2, 3, 4, 5, 6]
lista2 = [4, 5, 6, 7, 8, 9]
comuns = {x for x in lista1 if x in lista2}
print(f"Números comuns: {comuns}")
````

### ⚠️ Quando NÃO usar List Comprehensions

List comprehensions são ótimas para operações **simples e diretas**. Se a lógica ficar complexa, use um loop `for` normal para manter a legibilidade.

❌ **Ruim** (complexo demais):
```python
resultado = [x**2 if x % 2 == 0 else x**3 for x in range(100) if x % 3 != 0]

````python
resultado = []
for x in range(100):
    if x % 3 != 0:
        if x % 2 == 0:
            resultado.append(x**2)
        else:
            resultado.append(x**3)
````

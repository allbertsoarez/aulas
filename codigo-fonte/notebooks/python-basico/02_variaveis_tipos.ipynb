  # 📦 Variáveis e Tipos de Dados
  Em Python, não precisamos declarar o tipo da variável. O próprio Python descobre o tipo no momento em que você atribui um valor (Tipagem Dinâmica).

```python
# Tipos primitivos básicos
idade = 25                  # int (inteiro)
altura = 1.75               # float (ponto flutuante/decimal)
nome = "Maria"              # str (string/texto)
eh_estudante = True         # bool (booleano: True ou False)

# Usando a função type() para descobrir o tipo de cada variável
print(f"Idade: {idade} -> Tipo: {type(idade)}")
print(f"Altura: {altura} -> Tipo: {type(altura)}")
print(f"Nome: {nome} -> Tipo: {type(nome)}")
print(f"Estudante? {eh_estudante} -> Tipo: {type(eh_estudante)}")
```

  # 🛠️ Funções
  Funções permitem agrupar código para ser reutilizado. Usamos a palavra-chave `def`.

```python
# Função que NÃO retorna nada (apenas executa uma ação)
def saudacao(nome):
  print(f"Olá, {nome}! Bem-vindo ao Colab.")

# Função que RETORNA um valor (pode ser guardado em uma variável)
def calcular_area_retangulo(base, altura):
  area = base * altura
  return area

# Chamando as funções
saudacao("Alberto")

minha_area = calcular_area_retangulo(5, 10)
print(f"A área do retângulo é: {minha_area}")
```

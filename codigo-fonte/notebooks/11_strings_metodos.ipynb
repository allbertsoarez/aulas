# 🔤 Métodos de Strings

Strings em Python são **objetos poderosos** com dezenas de métodos úteis. Dominar esses métodos é essencial para:
- Limpar dados de formulários
- Processar textos de APIs
- Validar entradas do usuário
- Fazer web scraping

```python
texto = "  Olá, Mundo!  "

# Métodos de transformação
print(f"Original: '{texto}'")
print(f"upper(): '{texto.upper()}'")           # MAIÚSCULAS
print(f"lower(): '{texto.lower()}'")           # minúsculas
print(f"strip(): '{texto.strip()}'")           # remove espaços nas extremidades
print(f"lstrip(): '{texto.lstrip()}'")         # remove espaços à esquerda
print(f"rstrip(): '{texto.rstrip()}'")         # remove espaços à direita
print(f"replace('Mundo', 'Python'): '{texto.strip().replace('Mundo', 'Python')}'")
```

```python
# Métodos de busca e verificação
frase = "Python é uma linguagem poderosa"

print(f"startswith('Python'): {frase.startswith('Python')}")  # True
print(f"endswith('poderosa'): {frase.endswith('poderosa')}")  # True
print(f"find('linguagem'): {frase.find('linguagem')}")        # 15 (posição)
print(f"count('a'): {frase.count('a')}")                      # 4
print(f"'Python' in frase: {'Python' in frase}")              # True
```

```python
# Métodos de divisão e junção (MUITO IMPORTANTES!)

# split() - divide uma string em uma lista
csv_line = "Ana,20,Ciência da Computação,9.5"
dados = csv_line.split(",")
print(f"Split: {dados}")
print(f"Nome: {dados[0]}, Curso: {dados[2]}")

# join() - junta uma lista em uma string (o inverso do split)
palavras = ["Python", "é", "incrível"]
frase_junta = " ".join(palavras)
print(f"\nJoin: '{frase_junta}'")

# Usando separadores diferentes
print("-".join(palavras))  # Python-é-incrível
print(", ".join(palavras)) # Python, é, incrível
```

```python
# Métodos de verificação de tipo
testes = ["123", "abc", "ABC", "Abc123", "   ", ""]

print(f"{'Texto':<10} | {'isdigit':<8} | {'isalpha':<8} | {'isalnum':<8} | {'isspace':<8}")
print("-" * 55)
for t in testes:
    print(f"{f'{t}':<10} | {str(t.isdigit()):<8} | {str(t.isalpha()):<8} | {str(t.isalnum()):<8} | {str(t.isspace()):<8}")
```

```python
# 🎯 Caso de uso real: limpeza de dados
dados_sujos = [
    "  Ana Silva  ",
    "BRUNO SOUZA",
    "carla.oliveira@email.com",
    " (11) 99999-8888 ",
    "  CEP: 01001-000  "
]

print("=== Dados Limpos ===")
for dado in dados_sujos:
    # Remove espaços, converte para título
    limpo = dado.strip().title()
    print(f"'{limpo}'")
```

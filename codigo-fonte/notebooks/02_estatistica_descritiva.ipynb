  # 📊 Estatística Descritiva Aplicada
  
  Aqui aplicamos os conceitos de **Tendência Central** e **Dispersão** vistos na apostila de Estatística, utilizando o poder de cálculo do `numpy` e `pandas`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Usando os mesmos dados do notebook anterior
precos = [29.90, 39.90, 49.90, 69.90, 24.90, 89.90, 35.00, 42.50]

# Cálculos Estatísticos
media = np.mean(precos)
mediana = np.median(precos)
desvio_padrao = np.std(precos, ddof=1) # ddof=1 para desvio padrão amostral (n-1)

print(f"Média dos preços: R$ {media:.2f}")
print(f"Mediana dos preços: R$ {mediana:.2f}")
print(f"Desvio Padrão: R$ {desvio_padrao:.2f}")
```

  ### 📈 Visualização da Distribuição (Histograma)
  O histograma nos permite ver a "forma" da distribuição dos dados, complementando os números da média e desvio padrão.

```python
# Configurando o estilo do gráfico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# Criando o histograma
sns.histplot(precos, bins=5, kde=True, color='skyblue')

# Adicionando linhas de média e mediana
plt.axvline(media, color='red', linestyle='dashed', linewidth=2, label=f'Média (R$ {media:.2f})')
plt.axvline(mediana, color='green', linestyle='dashed', linewidth=2, label=f'Mediana (R$ {mediana:.2f})')

plt.title('Distribuição dos Preços dos Livros')
plt.xlabel('Preço (R$)')
plt.ylabel('Frequência')
plt.legend()
plt.show()
```

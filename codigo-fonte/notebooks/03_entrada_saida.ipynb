  # ⌨️ Entrada e Saída de Dados
  - `print()`: Exibe informações na tela.
  - `input()`: Lê informações digitadas pelo usuário.
  
  *Atenção: O `input()` sempre retorna um **texto (string)**. Se quiser um número, precisamos converter!*

```python
# Lendo dados do usuário (ao executar, uma caixinha aparecerá para você digitar)
nome = input("Digite o seu nome: ")
idade_texto = input("Digite a sua idade: ")

# Convertendo o texto para número inteiro
idade = int(idade_texto)

# Calculando o ano de nascimento
ano_atual = 2024
ano_nascimento = ano_atual - idade

# Exibindo o resultado usando f-strings (a forma mais moderna de formatar texto)
print(f"\nOlá, {nome}!")
print(f"Você tem {idade} anos e nasceu por volta de {ano_nascimento}.")
```

  # 🔁 Estruturas de Repetição
  Usamos laços para repetir um bloco de código várias vezes sem precisar reescrevê-lo.

```python
# --- LAÇO FOR ---
# O range(inicio, fim) gera uma sequência de números. O 'fim' é exclusivo.
print("Contando de 1 a 5:")
for i in range(1, 6):
  print(i)

print("\n--- Tabuada ---")
numero = int(input("Digite um número para a tabuada: "))
for i in range(1, 11):
  print(f"{numero} x {i} = {numero * i}")
```

```python
# --- LAÇO WHILE ---
# Repete ENQUANTO a condição for verdadeira
print("\n--- Contagem Regressiva ---")
contador = 5
while contador > 0:
  print(contador)
  contador -= 1  # Equivale a: contador = contador - 1
  print("Foguete lançado! 🚀")
```
